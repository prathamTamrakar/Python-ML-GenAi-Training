```python
import seaborn as sns
```

```python
df = sns.load_dataset('penguins')
```

```python
df.sample(6)
```

# Heatmap()

```python
# calculate corealtion matrix 

corealtion_matrix = df.corr(numeric_only=True)
#  Heatmap

sns.heatmap(corealtion_matrix)
```

```python
sns.heatmap(corealtion_matrix , annot = True)
```

```python
sns.heatmap(corealtion_matrix , annot = True, fmt = '.1f')
```

```python
sns.heatmap(corealtion_matrix , annot = True, linewidths=2, linecolor="b" )
```

```python
sns.heatmap(corealtion_matrix , annot = True, linewidths=2,cmap='winter')
# winter jet cool warm
```
