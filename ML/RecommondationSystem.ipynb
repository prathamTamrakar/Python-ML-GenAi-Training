```python
import ast
```

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i["name"])
    return L
```

```python
movies.iloc[19].genres
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
cast = movies['cast']
def convertcast(cast):
    count = 0
    L = []
    for i in ast.literal_eval(cast):
        if count!=3:
            L.append(i["name"])
            count = count + 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convertcast)
```

```python
movies.head()
```

```python
movies['crew'][0]
```

```python
crew = movies['crew']
def convertcrew(crew):
    s = ""
    for i in ast.literal_eval(crew):
            if i['job']=='Director':
                s = i['name']
    return s
```

```python
movies['crew'] = movies['crew'].apply(convertcrew)
```
