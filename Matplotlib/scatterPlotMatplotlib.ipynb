```python
x = ["python","C","C++","Java"]
y = [87,62,50,80]
```

```python
#bar-vertical
plt.bar(x,y)
plt.show()
```

```python
#bar-vertical
plt.bar(x,y)
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.show()
```

```python
plt.bar(x,y ,color="yellow" ,align="center" )
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.show()
```

```python
plt.bar(x,y,width=0.2 )
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.show()
```

```python
plt.bar(x,y,width=0.2,align="center" )
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.show()
```

```python
#edge color line width line style parameter
plt.bar(x,y ,color="blue",width=0.4,edgecolor='r',linewidth=3,linestyle='-' )
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.show()
```

```python
#color shade alpha parameter
plt.bar(x,y ,color="yellow",alpha=0.7) #color shade dark and low
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.show()
```

```python
#bar-horizontal graph
plt.barh(x,y ,color="yellow",label='popularity' )#label uses legend as a parameter
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.legend()
plt.show()
```

```python
#bar-horizontal graph
plt.barh(x,y ,color="yellow",label='popularity',height=0.4 )#label uses legend as a parameter
plt.title("programing language vs popularity")
plt.xlabel("language")
plt.xlabel("popularity")
plt.legend()
plt.show()
```
