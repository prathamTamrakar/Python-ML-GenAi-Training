```python
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df.sample(10)
```

1. Relationship Between Petal Length and Petal Width Across Species
- Plot the relationship between petal_length and petal_width for different species.
- Use axis-level (sns.scatterplot) and figure-level (sns.relplot) versions.
- Highlight species differences using hue='species'.

```python
sns.relplot(data=df,x="petal_length",y="petal_width",hue="species")
```

```python
sns.scatterplot(data=df,x="petal_length",y="petal_width",hue="species")
```

in the above plot ,
setosa species have minimum petal length and petal width, virginica have lagest petal length and petal width and versicolor have petal length and petal width in between the setosa and viginica

2. Trend of Sepal Length by Species
- Show how average sepal_length varies across species.

```python
sns.lineplot(data=df,y='sepal_length',x='species',errorbar='sd',marker='.')
```
