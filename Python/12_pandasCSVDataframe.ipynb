```python
import pandas as pd
```

Creating Dataframe

```python
# using lists

data_sample = [
    [50,60,47],
    [70,73,75],
    [100,400,34],
    [60,5,20]
]
data = pd.DataFrame(data_sample,columns=['Python','C++','Java'])
print(data)
```

```python
# using dicts
Data_dicts = {
    'col1':['name1','name2','name3','name4','name5','name6'],
    'col2':[100,34,234,55,344,66],
    'col3':[10,3,23,5,34,6],
    'col4':[1002,342,2342,552,3442,662],
}
students = pd.DataFrame(Data_dicts)
print(students)
students.set_index('col1',inplace=True)
print(students)
```

```python
# using read_CSV
weather = pd.read_csv("./Data/austin_weather.csv")
print(weather)
```
