WAP which will print count of even and odd numbers in list

```python
lit = [1,2,3,4,5]
even = odd = 0
for i in range (len(lit)):
    if(i%2==0):
        even+=1
    else:
        odd+=1
print("even",even)
print("odd",odd)
```

WAP to convert even digit to string even and if digit is odd then add 5 to it

```python
lit = [1,2,3,4,5]
for i in range (len(lit)):
    if(i%2==0):
        lit[i] = "even"
    else:
        lit[i] = lit[i]+5
print(lit)
```
