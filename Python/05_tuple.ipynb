```python
# it is immutable
tpl = ()
tp = tuple() # empty tuple
t = (1,2,4,55,"Hello")
print(tpl,tp,t)
```

```python
tpl=(5)
print(type(tpl)) # int
tpl=(5,)
print(type(tpl)) # tuple
tpl=5,4,3,5
print(type(tpl)) # tuple
tpl=5,
print(type(tpl)) # tuple
```

```python
tpl = (1,2,3,4,5,6,6,7,8,8,9,7,6,5,4,4,3,3)
print(tpl.index(4)) # 3
print(tpl.count(6)) # 3
```

```python
tpl = (1,2,3,4,5)
print(id(tpl))
print(tpl)
tpl+=(6,7,8)
print(id(tpl))
print(tpl)
tpl+=(9,10)
print(id(tpl))
print(tpl)
```

```python
tpl = (1,2,3,4,5,6)
lst=list(tpl)
for i in lst:
    if i%2!=0:
        lst.remove(i)
x=tuple(lst)
print(x,type(x))
```

```python
x = (1,2,3,4,5,6)
x=list(x)
lst = [i for i in x if i%2==0]
x=tuple(x)
print(x,type(x))
```
