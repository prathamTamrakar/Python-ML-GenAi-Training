```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,3,4,5,6,7,8] # x-data
y = [2,8,4,16,10,12,6,12,8] #y-data

plt.plot(x,y)
```

## Add using

- plt.xlabel()
- plt.ylabel()
- plt.show



```python
plt.plot(x,y)
plt.title("Numbers Count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.show # show plot
```

```python
#givin color to the line parameter
plt.plot(x,y,color="red")
plt.title("Numbers Count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.show # show plot
```

```python
#line style change dot 
# plt.plot(x,y ,color="black",linestyle='dashdot')
plt.plot(x,y ,color="black",linestyle='dotted')
plt.title("Numbers Count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.show # show plot
```

```python
#line width
plt.plot(x,y ,color="black",linestyle='dashdot' ,linewidth=3)
plt.title("Numbers Count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.show # show plot
```

```python
#line markesr in plot
plt.plot(x,y ,color="black",linestyle='dashdot' ,marker="*",markersize=20)
# plt.plot(x,y ,color="black",linestyle='dashdot' ,marker="d",markersize=20)

plt.title("Numbers Count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.show # show plot
```

```python
plt.plot(x,y ,color="black",linestyle='dashdot' ,marker="*",markersize=20,linewidth=2)
plt.title("Numbers Count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.show
```

```python
# multiline graph
x=[1,2,3,4,5,6,7,8,9] #x-data
y=[2,4,6,8,9,6,5] #y-data
z=[3,4,5,6,7,2,8] #z-data
plt.plot(x,y ,color="black",linestyle='dashdot' ,marker=">")
plt.title("Numbers count Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.plot(y,z ,color="red",linestyle='dashdot' ,marker=">")
plt.plot(x,z,color="blue",linestyle='dashdot' ,marker=">")
plt.show
```

```python
#label-
# legend()
plt.plot(x,y ,label="abc")
plt.plot(x,z, label="xyz")
plt.title("Numbers COUNT Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.legend()
plt.show
```

```python
#using grid backg
plt.plot(x,y ,label="abc")
plt.plot(x,z, label="xyz")
plt.title("Numbers COUNT Frequency")
plt.xlabel('Numbers')
plt.ylabel("Frequncy")
plt.legend()
plt.grid()
plt.show
```
