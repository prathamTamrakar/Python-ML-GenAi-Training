WAP which will take a number as a input sum of digit in the number

```python
n = int(input("Enter any number"))
sum=0
while n!=0:
    num=n%10
    n=n//10
    sum+=num
print(sum)
```

WAP which will take input from the user and give LCM of both the number

```python
a = int(input('Enter any number'))
b = int(input('Enter any number'))
for i in range (a,a*b+1):
    if i%a==0 and i%b==0:
        print("LCM is",i)
        break;
```

WAP which will add all the number between 10 to 55 but it will skip all the number which is divisible by 3 and 5

```python
sum =0
for i in range (10,55):
    if i%3!=0 and i%5!=0:
        sum = sum + i
print(sum)
```

WAP which will print multiplication of all the even and odd of given factor

```python
n = int(input("Enter any number"))
i=1
evenMul=1
oddMul=1
for i in range (1,n+1):
    if n%i==0:
        if(i%2==0):
            evenMul*=i
        else:
            oddMul*=i
print("evenMul",evenMul)
print("oddMul",oddMul)
```
