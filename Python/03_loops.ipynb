WAP to check palindrome

```python

n=int(input("Enter the number :"))
temp=n
rev=0
while(n!=0):
    rem=n%10
    rev=rev*10+rem
    n=n//10
if(rev==temp):
    print("Palindrome")
else:
    print("Not a Palindrome")
```

WAP to check for Armstrong number, narcissistic number or pluperfect digital invariant

```python
n=int(input("Enter the number :"))
sum=0
temp=n
m=n
count=0
while(m!=0):
    m=m//10
    count+=1

while(n!=0):
    rem=n%10
    sum=sum+rem**count
    n=n//10
if(sum==temp):
    print("Armstrong number")
else:
    print("Not an armstrong number")
```

WAP which take two number as input and find the LCM of both the number

```python
n1=int(input("Number 1:"))
n2=int(input("Number 2:"))
i=max(n1,n2)
while(i<=n1*n2):
    if(i%n1==0 and i%n2==0):
        print("LCM is ",i)
        break
    i+=1
```

```python
n1=int(input("Number 1:"))
n2=int(input("Number 2:"))
i=min(n1,n2)
while(i<=n1 and i<=n2):
    if(n1%i==0 and n2%i==0):
        print("HCF is ",i)
        break
    i-=1
```
