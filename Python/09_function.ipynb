```python
n= int(input("Enter the value of n"))
r = int(input("Enter the value of r"))
p=n-r
nFact = 1
rFact = 1
pFact = 1

for i in range(n,0,-1):
    nFact*=i

for i in range(r,0,-1):
    rFact*=i

for i in range(p,0,-1):
    pFact*=i

combination = (nFact/(rFact*pFact))
print(combination)
```

```python
n= int(input("Enter the value of n"))
r = int(input("Enter the value of r"))
p=n-r
def fact(num):
    for i in range(num,0,-1):
        fact*=i
    

nFact = fact(n)
rFact = fact(r)
pFact = fact(n-r)

combination = (nFact/(rFact*pFact))
print(combination)
```
