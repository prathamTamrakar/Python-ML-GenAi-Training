```python
n= int(input("Enter the value of n"))
r = int(input("Enter the value of r"))
p=n-r
nFact = 1
rFact = 1
pFact = 1

for i in range(n,0,-1):
    nFact*=i

for i in range(r,0,-1):
    rFact*=i

for i in range(p,0,-1):
    pFact*=i

combination = (nFact/(rFact*pFact))
print(combination)
```

```python
n= int(input("Enter the value of n"))
r = int(input("Enter the value of r"))
p=n-r
def fact(num):
    fact = 1
    for i in range(num,0,-1):
        fact*=i
    return fact
    

nFact = fact(n)
rFact = fact(r)
pFact = fact(n-r)

combination = (nFact/(rFact*pFact))
print(combination)
```

WAP to implement sum of all the prime numbers in the list

```python
def primeNum(num):
    for i in range (2,num):
        if num%i==0:
            return False
    return True

x = [5,3,2,5]
sum = 0
for i in x:
    if(primeNum(i)):
        sum+=i
    else:
        continue
print(sum)
```

```python
# filter
# filter(function,sequence)

# make a list of all the element greater then 4
lst = [2,3,4,4,5,6,6,7,78,8,85,4,43]
var = list(filter(lambda x: x>4,lst))
print(var)
```

```python
# make a list of all the element which starting from a
lst = ["abx","ghf","Asd","hgf"]
var = list(filter(lambda x: x[0]=='a' or x[0]=="A",lst))
print(var)
```

```python
from functools import reduce
lst = [3,4,6,4,3,6,8,5,4,6,8,4,3,24,6,8,6,53,33,45,6,77,4,4,6,76]
var1 = reduce(lambda x,y:max(x,y),lst)
var2 = reduce(lambda x,y:x>y,lst)
var3 = reduce(lambda x,y:x if x>y else y,lst)
var4 = reduce(lambda x,y:min(x,y),lst)
print(var1)
print(var2)
print(var3)
print(var4)
```
