```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [10,20,30,40,15]
l = ["C","C++","Python","java","R"]
```

```python
plt.pie(x)
plt.show()
```

```python
plt.pie(x,labels=l)
plt.show()
```

```python
ex= [0.3,0,0,0.2,0]
plt.pie(x,labels=l,explode=ex)
plt.show()
```

```python
c= ['r','g','y','b','m']
plt.pie(x,labels=l,colors=c)
plt.show()
```

```python
plt.pie(x,labels=l,colors=c,autopct="%0.1f%%",shadow=True)
plt.show()
```
