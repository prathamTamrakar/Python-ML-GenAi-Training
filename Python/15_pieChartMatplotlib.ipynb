```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [10,20,30,40,15]
l = ["C","C++","Python","java","R"]
```

```python
plt.pie(x)
plt.show()
```

```python
plt.pie(x,labels=l)
plt.show()
```

```python
ex= [0.3,0,0,0.2,0]
plt.pie(x,labels=l,explode=ex)
plt.show()
```

```python
c= ['r','g','y','b','m']
plt.pie(x,labels=l,colors=c)
plt.show()
```

```python
plt.pie(x,labels=l,colors=c,autopct="%0.1f%%",shadow=True)
plt.show()
```

```python
ex= [0.3,0,0,0.2,0]
plt.pie(x,labels=l,explode=ex,radius=0.5)
plt.show()
```

```python
plt.pie(x,labels=l,labeldistance=0.5)
plt.show()
```

```python
ex= [0.3,0,0,0,0]
plt.pie(x,labels=l,autopct="%0.1f%%",explode=ex,startangle=180)
plt.show()
```

```python
x = [10,20,30,40]
x1 = [60,50,30,20]
y = ['C','C++','Java',"Python"]
c = ['r','g','b','m']
plt.pie(x,labels=y,radius=1.5)
plt.pie(x1,radius=0.7,colors=c)
plt.show()
```

```python
x = [10,20,30,40]
y = ['C','C++','Java',"Python"]
plt.pie(x,labels=y)
plt.pie([1],colors='w',radius=0.95)
plt.show()
```
