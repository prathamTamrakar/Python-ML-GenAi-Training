```python
# creating multiple combination in dictionary

#List of list
d1 = dict([{1,'abc'},{2,'cde'},{3,'frd'}])
print(d1)

#tuple of list
d2 = dict((['Man',90],['woman',90]))
print(d1)

# tuple of tuple
d3 = dict([(1,'a'),(2,'c'),(3,'f')])
print(d3)

#tuple of set
d4 = dict(({'mango',50},{'orange',80}))
print(d4)

#set of tuple 
d5 = dict({((5,6,7),'a'),(3,'b')})
print(d5)
```

```python
x = [3,5,3,7,9,5,8,3,7,9,3,5]
dic = {}
for i in x:
    count =  0
    for j in x:
        if i==j:
            count+=1
    dic[i] = count
print(dic)
```

```python
x = [3,5,3,7,9,5,8,3,7,9,3,5]
dic = {}
for i in x:
    dic[i] = x.count(i)
print(dic)
```

DELETE OR REMOVE ELEMENT FROM DICTIONARY

```python
my_dict = {'name':'satish','age':20,'address':'JBP'}
print(my_dict)
print(my_dict.pop('address'))
print(my_dict)
```

```python
my_dict = {'name':'satish','age':20,'address':'JBP'}
print(my_dict)
print(my_dict.popitem())
print(my_dict)
```

```python
my_dict = {'name':'satish','age':20,'address':'JBP'}
print(my_dict)
del my_dict['address']
print(my_dict)
del my_dict
print(my_dict)
```

```python
my_dict = {'name':'satish','age':20,'address':'JBP'}
print(my_dict)
my_dict.clear()
print(my_dict)
```

```python
x = [3,4,5,6,7]
y=x # shallow copy
x[3] = 15
y[3] = 14
print(x)
print(y)
```

```python
x = [3,4,5,6,7]
y=x.copy() 
x[3] = 15
y[3] = 14
print(x)
print(y)
```

```python
squares = {2:4,3:9,4:16,5:25}
my_dict = squares.copy()
squares[6] = 4
print(squares)
print(my_dict)
```

```python
x = {3:7,8:9,12:25,13:37}

for k,v in x.items():
    if v% 2==0:
        x[k]==5
    else:
        x[k]-=5
print(x)
```

```python
x = {3:[1,2,3,4],4:[7,8,9,10],5:[2,3,4,7,8],6:[7,3,5,2]}
for k,v in x.items():
    count = 0
    for i in v:
        if i%2==0:
            count+=1
    x[k] = count
print(x)
```

```python
x = {3:[1,2,3,4],4:[7,8,9,10],5:[2,3,4,7,8],6:[7,3,5,2]}
for k,v in x.items():
    for i in range (0,len(v)):
        v[i] = v[i] + 1
print(x)
```

```python
x = {3:[1,2,3,4],4:[7,8,9,10],5:[2,3,4,7,8],6:[7,3,5,2]}
for k,v in x.items():
    x[k] = [i+1 for i in v]
print(x)
```

```python
d = {'a':1,'b':2,'c':3,'d':4}
new_dict = {k:v for k,v in d.items() if v>2 }
print(new_dict)
```

```python
d = {'a':1,'b':2,'c':3,'d':4}
new_dict = {k+'c':v*v for k,v in d.items() if v>2 }
print(new_dict)
```
