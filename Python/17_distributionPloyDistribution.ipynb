```python
import seaborn as sns
```

# Distribution PLot

- figure level  -> displot
- axix Level -> histplot, rugplot, kdeplot

```python
df = sns.load_dataset('penguins')
```

```python
df.head()
```

```python
sns.histplot(data=df,x="body_mass_g", color="yellow")
# univarient analysis
```

```python
sns.histplot(data=df,x="bill_length_mm", color="yellow")
```

```python
sns.histplot(data=df,x="body_mass_g", color="yellow",bins=15)
```

```python
sns.histplot(data=df,x="body_mass_g", color="yellow",bins=15)
```

```python
sns.displot(kind='hist',data=df,x="body_mass_g",bins=15)
```

```python
sns.displot(kind="hist",data=df,x="species")
```

```python
sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species")

```

```python
sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="island")
```

```python
sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="sex")
```

```python
# sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="step")
# sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="bars")
sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="poly")

```

```python
# sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="poly",col="island")
sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="poly",col="sex")

```

```python
# sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="poly",row="sex")
sns.displot(kind="hist",data=df,x="flipper_length_mm",hue="species",element="poly",row="island")

```

a. axix level - kde

```python
sns.kdeplot(data=df,x="flipper_length_mm")
```

b. figure level - dist for kde

```python
sns.displot(kind="kde",data=df,x="flipper_length_mm")
```

```python
sns.displot(kind="kde",data=df,x="flipper_length_mm",hue="sex",fill= True)
```

# Bivariate using kdeplot

```python
sns.displot(kind="kde",data=df,x="flipper_length_mm",y="bill_length_mm")
```

rug -> supporing plot

```python
sns.kdeplot(data=df,x="flipper_length_mm")
sns.rugplot(data=df,x="flipper_length_mm")
```
